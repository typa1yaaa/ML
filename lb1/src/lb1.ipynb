#### 2.3 Рассчитаем характеристики, полученные методом describe (п. 1.3) с использованием методов NumPy. 

```python
def describe_numpy(data_numpy):
    desc = {}
    desc['count'] = data_numpy.shape[0]
    desc['mean'] = np.mean(data_numpy, axis=0)
    desc['std'] = np.std(data_numpy, axis=0, ddof=1) # ddof=1 - несмещенная оценка, делит на (n-1)
    desc['min'] = np.min(data_numpy, axis=0)
    desc['25%'] = np.percentile(data_numpy, 25, axis=0)
    desc['50%'] = np.median(data_numpy, axis=0)
    desc['75%'] = np.percentile(data_numpy, 75, axis=0)
    desc['max'] = np.max(data_numpy, axis=0)

    return desc

desc = describe_numpy(data_numpy)

for key, value in desc.items():
    print(f"{key}: {value}")
```

**OneHotEncoder** преобразует категориальную метку в бинарный вектор. reshape(-1, 1) нужен, чтобы OneHotEncoder понимал, что нужно одну колонку категориальных меток превратить в бинарные вектора.

#### 3.2. Для датафрейма (из п. 1.4), получим все столбцы признаков (столбцы не содержащие метки классов). Преобразуем полученные столбцы в массив NumPy. 

```python
data_only_var = data_pandas.drop(columns=['Label', 'Unnamed: 0'])
print(data_only_var.head(), '\n')

data_numpy_array = np.array(data_only_var)
print(data_numpy_array.shape)
print(data_numpy_array[:5])
```

#### 3.3. Для массива NumPy (из п. 3.2) применим StandardScaler, MinMaxScaler, MaxAbsScaler и RobustScaler. Для каждого из результатов построим гистограммы по каждому признаку без разделения по классам. 

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler

scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'MaxAbsScaler': MaxAbsScaler(),
    'RobustScaler': RobustScaler()
}

scaled_data = {}

for name, scaler in scalers.items():
    scaled_data[name] = scaler.fit_transform(data_numpy_array)
```

```python
for name, data_scaled in scaled_data.items():
    n_features = data_scaled.shape[1]
    
    fig, axs = plt.subplots(1, n_features, figsize=(4*n_features, 4))
    fig.suptitle(name, fontsize=16)
    
    for i in range(n_features):
        sns.histplot(data_scaled[:, i], bins=15, kde=True, ax=axs[i], color='skyblue')
        axs[i].set_title(f'Var{i+1}')
    
    plt.show()
```

**StandartScaler:** X_scaled = (X - mean) / std

Центрирует данные (среднее 0) и нормализирует дисперсию (std = 1), но чувствителен к выбросам.

**MinMaxScaler:** X_scaled = (X - min) / (max - min)

Сжимает данные в диапазон [0, 1] и сохраняет форму распределения, но выбросы сжимают основную массу данных.

**MaxAbsScaler:** X_scaled = X / max(|X|)

Масштабирует данные в диапазон [-1, 1], деля каждый признак на его
максимальное абсолютное значение. Не сдвигает данные (не центрирует), поэтому подходит для разреженных данных.

**RobustScaler:** X_scaled = (X - median) / IQR

Использует медиану и межквартильный размах (IQR = Q75 - Q25) вместо среднего и стандартного отклонения. Устойчив к выбросам, так как медиана и IQR менее чувствительны к экстремальным значениям.

Сравним все scaler-ы и проверим выбросы:

```python
fig, axs = plt.subplots(1, len(var_cols), figsize=(5 * len(var_cols), 5))

for i in range(len(var_cols)):
    for name, data_scaled in scaled_data.items():
        sns.kdeplot(data_scaled[:, i], ax=axs[i], label=name, lw=2)
    axs[i].set_title(f'Var{i+1}')
    axs[i].legend(fontsize=8)

plt.suptitle('Сравнение scaler (ядерная оценка плотности)', fontsize=14)
plt.tight_layout()
plt.show()

fig, axs = plt.subplots(1, len(var_cols), figsize=(4 * len(var_cols), 4))
for i, col in enumerate(var_cols):
    sns.boxplot(y=data_pandas[col], ax=axs[i])
    axs[i].set_title(col)
plt.suptitle('Оценка выбросов', fontsize=14)
plt.tight_layout()
plt.show()
```

Видим, что только у Var4 есть незначительный выброс, но в остальном все сотальные данные имеют достаточно симметричные распределения.

Поскольку у данных выбросов практически нет, а рпспределения относительно симметричны, лучше всего подходит **StandartScaler** - он центрирует данные и нормализует дисперсию. **RobustScaler** дал бы аналогичный результат, но, поскольку его используют для устойчивости к выбросам, с этими данными он не актуален. **MinMaxScaler** и **MaxAbsScaler** слишком сжимают данные.

### 4. Понижение размерности:

#### 4.1. Для набора данных применим понижение размерности, используя PCA и TSNE из Sklearn.

Оценим для PCA необходимое количество компонент для сохранения 90% информации:

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

pca = PCA()
pca.fit(data_numpy_array)

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

n_components_90 = np.argmax(cumulative_variance >= 0.9) + 1

print("число компонент:", n_components_90)
```

Построим диаграммы рассеяния для PCA и TSNE:

```python
pca_2d = PCA(n_components=2)
data_pca = pca_2d.fit_transform(data_numpy_array)

plt.figure()
sns.scatterplot(x=data_pca[:,0], y=data_pca[:,1], hue=labels, palette='Set1')
plt.title("PCA")
plt.show()


tsne = TSNE(n_components=2)
data_tsne = tsne.fit_transform(data_numpy_array)

plt.figure()
sns.scatterplot(x=data_tsne[:,0], y=data_tsne[:,1], hue=labels, palette='Set1')
plt.title("TSNE")
plt.show()
```

**РСА:** Линейный метод, который сохраняет глобальную структуру и дисперсию. Заметим, что данные на графике PCA разделяются на три группы, которые соответствуют трем классам. Класс 1 и класс 2 хорошо отделены от остальных, в то время как класс 3 немного пересекается с другими классами. PCA смог выделить основные направления изменения данных и частично разделить классы, но между некоторыми классами все еще есть перекрытие.

**TSNE:** Нелинейный метод, которых лучше сохраняет локальные расстояния и кластерную структуру. На графике TSNE разделение классов видно гораздо лучше. Все три класса образуют три четкие группы и практически не пересекаются. TSNE лучше сохраняет структуру данных и расстояния между похожими объектами.

**Вывод:** Метод PCA хорошо уменьшает размерность и показывает общее распределение данных, но не всегда полностью корректно разделяет классы. Метод TSNE показывает более четкое разделение классов, он лучше подходит для визуализации структуры данных и выявления кластеров.
